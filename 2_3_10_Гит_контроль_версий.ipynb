В блокноте реализованы функции для:
1. генерации списка равномерно распределённых случайных целых чисел;
2. построения частотного ряда;
3. расчёта минимума, среднего, медианы, моды и максимума.

```python
import random
from collections import Counter
from statistics import mean, median, multimode
```

## 1. Генерация случайного списка

Функция возвращает список длины `length`. Каждое значение — случайное целое число от `a` до `b` включительно.

```python
def generate_random_list(a: int, b: int, length: int) -> list[int]:
    """Создаёт список случайных целых чисел от a до b включительно."""
    if not all(isinstance(value, int) and not isinstance(value, bool)
               for value in (a, b, length)):
        raise TypeError("Параметры a, b и length должны быть целыми числами")
    if a > b:
        raise ValueError("Параметр a не должен быть больше b")
    if length < 0:
        raise ValueError("Длина списка не может быть отрицательной")

    return [random.randint(a, b) for _ in range(length)]
```

## 2. Построение частотного ряда

Ключ словаря — значение из списка, значение словаря — количество его появлений. Результат сортируется по ключам.

```python
def build_frequency_dict(values: list[int]) -> dict[int, int]:
    """Возвращает частотный ряд для переданного списка."""
    return dict(sorted(Counter(values).items()))
```

## 3. Расчёт описательных статистик

Если наиболее часто встречающихся значений несколько, поле `mode` содержит их список.

```python
def calculate_statistics(values: list[int]) -> dict[str, object]:
    """Вычисляет минимум, среднее, медиану, моду и максимум."""
    if not values:
        raise ValueError("Список не должен быть пустым")

    modes = multimode(values)
    mode_result = modes[0] if len(modes) == 1 else modes

    return {
        "minimum": min(values),
        "mean": mean(values),
        "median": median(values),
        "mode": mode_result,
        "maximum": max(values),
    }
```

## Пример выполнения

```python
example_list = [1, 2, 3, 1, 2, 2]

print("Список:", example_list)
print("Частотный ряд:", build_frequency_dict(example_list))
print("Статистики:", calculate_statistics(example_list))
```

## Запуск со случайными данными

Значения `a`, `b` и `length` можно изменить. `random.seed(42)` нужен для воспроизводимости примера.

```python
random.seed(42)
random_list = generate_random_list(a=1, b=5, length=20)

print("Случайный список:", random_list)
print("Частотный ряд:", build_frequency_dict(random_list))
print("Статистики:", calculate_statistics(random_list))
```

## Проверка функций

```python
test_list = generate_random_list(1, 5, 20)
assert len(test_list) == 20
assert all(1 <= value <= 5 for value in test_list)
assert build_frequency_dict([1, 2, 3, 1, 2, 2]) == {1: 2, 2: 3, 3: 1}
assert calculate_statistics([1, 2, 3, 1, 2, 2]) == {
    "minimum": 1,
    "mean": 11 / 6,
    "median": 2.0,
    "mode": 2,
    "maximum": 3,
}

print("Все проверки успешно пройдены.")
```
